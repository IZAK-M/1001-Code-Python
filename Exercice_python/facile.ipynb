#### Exercice 3
Ecrire un programme en Python qui demande à l'utilisateur de saisir deux nombres a et b et de lui afficher leur maximum.

```python
# Sans utilisation de max()
a = float(input("Tapez la valeur du nombre a : "))
b = float(input("Tapez la valeur du nombre b : "))

if a > b:
    print(f"Valeur maximum : {a}")
else:
    print(f"Valeur maximum {b}")
```

```python
# Avec utilisation de max()
a = float(input("Tapez la valeur du nombre a : "))
b = float(input("Tapez la valeur du nombre b : "))

print(f"Le maximum est {max(a, b)}")
```

#### Exercice 7
Ecrire un programme en Python qui demande à l'utilisateur de saisir 3 nombre x, y et z et de lui afficher leur maximum

```python
# Avec une liste :
x = float(input("Tapez la valeur du nombre x : "))
y = float(input("Tapez la valeur du nombre y : "))
z = float(input("Tapez la valeur du nombre z : "))

list_nombre = [x,y,z]

print(f"Le maximum est : {max(list_nombre)}")
```

```python
# Avec un dictionnaire :
x = float(input("Tapez la valeur du nombre x : "))
y = float(input("Tapez la valeur du nombre y : "))
z = float(input("Tapez la valeur du nombre z : "))

if x > y and x > z:
    print(f"Le maximum est : {x}")
elif y > x and y > z:
    print(f"Le maximum est : {y}")
else:
    print(f"Le maximum est : {z}")
```

#### Exercice 8
Ecrire un programme en Python qui demande à l'utilisateur de saisir un nombre entier n et de lui afficher la valeur de la somme 1 + 2 + … + n =

```python
n = int(input("Tapez le valeur nombre n : "))

somme = 0

for i in range (1, n+1):
    somme += i

print(f"La somme 1+2+...+n = {somme}")
```

#### Exercice 9
Ecrire un programme en Python qui demande à l'utilisateur de saisir un nombre entier n et de lui afficher n ! (Factoriel)

```python
n = int(input("Tapez la valeur du nombre n : "))

factoriel = 1

for i in range (1, n+1):
    factoriel *= i

print(f"Le {n}! = {factoriel}")
```

```python
# Fonction pour calculer le factoriel d'un nombre
def factoriel(n):
    if n == 0:
        return 1
    else:
        return n * factoriel(n-1)
    
# --- Utilisation de la fonction factoriel ---

n = int(input("Entrez un nombre entier : "))
 
# Calculer le factoriel de n en appelant la fonction factoriel
resultat = factoriel(n)
 
# Afficher le résultat
print(f"Le {n}! = {resultat}")
```
